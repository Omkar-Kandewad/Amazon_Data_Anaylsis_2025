```python
# Line Chart- Monthly Sales 
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
monthly_sales=df.groupby("Month")["Total Sales"].sum().reindex(month_order)
```

```python
monthly_sales
```

```python
plt.figure(figsize=(15,8))
plt.plot(monthly_sales.index.unique(),monthly_sales.values,marker="o",linewidth=2.5,markersize=8,alpha=1)
plt.xlabel("Month",size=20)
plt.ylabel("Total Sales",size=20)
plt.title("Monthly Sales",size=20)
plt.xticks(fontsize=13)
plt.yticks(fontsize=15)
plt.grid(True)
for x, y in zip(monthly_sales.index, monthly_sales.values):
    plt.text(x, y, int(y), ha='center', va='bottom',size=15)
plt.show()
```

```python
##line chart--> Monthly Quantity
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
monthly_quantity=df.groupby("Month")["Quantity"].sum().reindex(month_order)
```

```python
monthly_quantity
```

```python
plt.figure(figsize=(15,8))
plt.plot(monthly_quantity.index.unique(),monthly_quantity.values,marker="*",markersize=8,alpha=1)
plt.xlabel("Month",size=20)
plt.ylabel("Quantity",size=20)
plt.title("Monthly Quantity",size=20)
plt.xticks(fontsize=13)
plt.yticks(fontsize=15)
plt.grid(True)
for x, y in zip(monthly_quantity.index, monthly_quantity.values):
    plt.text(x, y, int(y), ha='center', va='bottom',size=15)
plt.show()
```

```python
# Horizontal bar cchar-->Revenue By Location
```

```python
location_sales = df.groupby('Customer Location')['Total Sales'].sum().sort_values()
plt.figure(figsize=(13,5))
plt.barh(location_sales.index, location_sales.values)
plt.title("Revenue by Location",size=20)
plt.xlabel("Total Sales",size=20)
plt.ylabel("Location",size=20)
plt.xticks(fontsize=13)
plt.yticks(fontsize=15)
for i, v in enumerate(location_sales.values):
    plt.text(v, i, int(v), va='center',size=15)
plt.show()
```

```python
# Pie Chart--> Status Mode
```

```python
status_counts = df['Status'].value_counts()
labels=["Cancelled","pending","Completed",]
plt.figure(figsize=(6,6))
plt.pie(status_counts.values,
        labels=status_counts.index,
        autopct='%1.1f%%',
        startangle=140)
plt.legend(labels,loc="lower right",fontsize=7,title="Status Mode",title_fontsize=8,shadow=True)
plt.title("Status Mode Distribution")
plt.show()
```

```python
# Pie Chart--> Payment Mode
```

```python
payment_counts = df['Payment Method'].value_counts()

plt.figure(figsize=(6,6))
labels=["Amazon Pay","Gift Card","PayPa","Debit Card","Credit Card"]
plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct='%1.1f%%',
        startangle=140)
plt.legend(labels,loc="lower right",fontsize=7,title="Payment Mode",title_fontsize=8,shadow=True)
plt.title("Payment Method Distribution")
plt.show()
```

```python

import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 2, figsize=(20, 18))
fig.suptitle("Amazon Sales Analysis Dashboard", fontsize=24, fontweight='bold')

# ------------------ 1️ Payment Method Pie ------------------
payment_counts = df['Payment Method'].value_counts()

axes[0, 0].pie(
    payment_counts.values,
    autopct='%1.1f%%',
    startangle=140
)
axes[0, 0].set_title("Payment Method Distribution", fontsize=16)

axes[0, 0].legend(
    payment_counts.index,
    loc="lower right",
    fontsize=9,
    title="Payment Mode",
    title_fontsize=10,
    shadow=True
)

# ------------------ 2️ Order Status Pie ------------------
status_counts = df['Status'].value_counts()

axes[0, 1].pie(
    status_counts.values,
    autopct='%1.1f%%',
    startangle=140
)
axes[0, 1].set_title("Order Status Distribution", fontsize=16)

axes[0, 1].legend(
    status_counts.index,
    loc="lower right",
    fontsize=9,
    title="Status",
    title_fontsize=10,
    shadow=True
)

# ------------------ 3️ Revenue by Location (Barh) ------------------
location_sales = df.groupby('Customer Location')['Total Sales'].sum().sort_values()

axes[1, 0].barh(location_sales.index, location_sales.values)
axes[1, 0].set_title("Revenue by Location", fontsize=16)
axes[1, 0].set_xlabel("Total Sales", fontsize=14)
axes[1, 0].set_ylabel("Location", fontsize=14)

for i, v in enumerate(location_sales.values):
    axes[1, 0].text(v, i, int(v), va='center', fontsize=12)

# ------------------ 4️ Monthly Quantity Trend ------------------
axes[1, 1].plot(
    monthly_quantity.index,
    monthly_quantity.values,
    marker="*",
    markersize=10,
    linewidth=2
)

axes[1, 1].set_title("Monthly Quantity Trend", fontsize=16)
axes[1, 1].set_xlabel("Month", fontsize=14)
axes[1, 1].set_ylabel("Quantity", fontsize=14)
axes[1, 1].grid(True)

for x, y in zip(monthly_quantity.index, monthly_quantity.values):
    axes[1, 1].text(x, y, int(y), ha='center', va='bottom', fontsize=12)

# ------------------ 5️ Monthly Sales Trend ------------------
axes[2, 0].plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    markersize=8,
    linewidth=2.5
)

axes[2, 0].set_title("Monthly Sales Trend", fontsize=16)
axes[2, 0].set_xlabel("Month", fontsize=14)
axes[2, 0].set_ylabel("Total Sales", fontsize=14)
axes[2, 0].grid(True)

for x, y in zip(monthly_sales.index, monthly_sales.values):
    axes[2, 0].text(x, y, int(y), ha='center', va='bottom', fontsize=12)

# ------------------ 6️ Empty / Summary Panel ------------------
axes[2, 1].axis('off')
axes[2, 1].text(
    0.1, 0.6,
    "Key Insights:\n"
    "- Digital payments dominate\n"
    "- Most orders are completed\n"
    "- Sales peak in certain months\n"
    "- Location-based revenue varies significantly",
    fontsize=14
)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()

```
